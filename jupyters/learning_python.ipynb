## 修饰符

```python
from typing import Callable

from adodbapi.examples.db_print import kw_args


def func_a(some_func: Callable) -> Callable:
    print(f"在调用函数{some_func}之前先打印该信息")
    return some_func

def func_b(a, b):
    return a + b

func_b(1, 2)
```

```python
@func_a
def func_c(a, b):
    return a + b

func_c(1, 2)
```

```python
def func_d(some_func: Callable) -> Callable:
    return lambda a, b: some_func(a, b) + 3

@func_d
def func_e(a, b):
    return a + b

func_e(1, 2)
```

## 函数参数

```python
# 位置参数不能出现在具名参数之前
def func(a, b, c, d=3, e):
    return a + b + c + d + e

func(1, 2, 3, 4, d=5)
```

```python
# 参数列表通过*运算符来强制用户进行具名传参
def func(a, b, *, d=3):
    return a + b + d
print(func(1, 2, 3))
print(func(1, 2, d=3))
```

```python
# 可变参数列表(位置参数)
def func(a, b, *args, d=3):
    print(args)
    return a + b + d + sum(args)

func(1, 2, 3, 4, 5, d=6)
```
